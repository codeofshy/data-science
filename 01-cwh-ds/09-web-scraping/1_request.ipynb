## we have reviewd the https://books.toscrape.com/
- with this we understand that when the pages change the int in the url changes
- total number of pages we have is 50
- page 1 : https://books.toscrape.com/catalogue/page-1.html
- page 5 : https://books.toscrape.com/catalogue/page-5.html
- page 50 : https://books.toscrape.com/catalogue/page-50.html
- let's scrape all the pages and store them in a books_html folder

```python
# testing for 1 page first
import requests

books = requests.get('https://books.toscrape.com/catalogue/page-1.html')
b = books.text 

# print(books.text)  # testing 
# print(f"Status of the books url : {books.status_code}")  # testing

print('Staring to scrape the pages from 1 to 50')

with open("page1.txt", "w") as f:
    f.write(b)

print("Scraped successfully...!")
```

```python
# Excercise
import requests

books = requests.get('https://books.toscrape.com/catalogue/page-1.html')

for i in range(1, 51):
    books = requests.get(f'https://books.toscrape.com/catalogue/page-{i}.html')
    with open(f"books_html/page{i}.html", "w") as f:
        f.write(books.text)
        print(f"Downloaded successfully, pg {i}.")

```
