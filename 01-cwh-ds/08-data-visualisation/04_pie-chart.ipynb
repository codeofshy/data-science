```python
import matplotlib.pyplot as plt
```

```python
players = ["Sachin", "Sehwag", "Kohli", "Yuvraj"]
runs_5yrs = [1200, 600, 300, 700]

plt.pie(runs_5yrs, labels=players)
plt.title("Runs in 5yrs")
plt.show()
```

```python
players = ["Sachin", "Sehwag", "Kohli", "Yuvraj"]
runs_5yrs = [1200, 600, 300, 700]

player_colors = ["skyblue", "purple", "yellow", "brown"]

plt.pie(runs_5yrs, labels=players, colors = player_colors)
plt.title("Runs in 5yrs")
plt.show()
```

```python
# https://matplotlib.org/stable/api/_as_gen/matplotlib.patches.Wedge.html
# by accessing the above url you can find the types of wedge props and there uses

players = ["Sachin", "Sehwag", "Kohli", "Yuvraj"]
runs_5yrs = [1200, 600, 300, 700]

player_colors = ["skyblue", "purple", "yellow", "brown"]

# wedgeprops
plt.pie(
    runs_5yrs, 
    labels=players, 
    colors = player_colors,
    wedgeprops={"edgecolor":"black", "linewidth":3, "linestyle":"--"}
)
plt.title("Runs in 5yrs")
plt.show()
```

```python
# changing edgeprops fill to False

players = ["Sachin", "Sehwag", "Kohli", "Yuvraj"]
runs_5yrs = [1200, 600, 300, 700]

player_colors = ["skyblue", "purple", "yellow", "brown"]

# wedgeprops
plt.pie(
    runs_5yrs, 
    labels=players, 
    colors = player_colors,
    wedgeprops={"edgecolor":"black", "linewidth":3, "linestyle":"--", "fill":False}
)
plt.title("Runs in 5yrs")
plt.show()
```

```python
# explode : pop a specefic section

expl = [0.1, 0, 0, 0]

players = ["Sachin", "Sehwag", "Kohli", "Yuvraj"]
runs_5yrs = [1200, 600, 300, 700]

player_colors = ["skyblue", "purple", "yellow", "brown"]

plt.pie(runs_5yrs, labels=players, explode= expl, colors = player_colors)
plt.title("Runs in 5yrs")
plt.show()
```

```python
expl = [0.1, 0.3, 0, 0]

players = ["Sachin", "Sehwag", "Kohli", "Yuvraj"]
runs_5yrs = [1200, 600, 300, 700]

player_colors = ["skyblue", "purple", "yellow", "brown"]

plt.pie(runs_5yrs, labels=players, explode= expl, colors = player_colors)
plt.title("Runs in 5yrs")
plt.show()
```

```python
expl = [0.1, 0.3, 0.2, 0]

players = ["Sachin", "Sehwag", "Kohli", "Yuvraj"]
runs_5yrs = [1200, 600, 300, 700]

player_colors = ["skyblue", "purple", "yellow", "brown"]

plt.pie(runs_5yrs, labels=players, explode= expl, colors = player_colors)
plt.title("Runs in 5yrs")
plt.show()
```

```python
# shadow
expl = [0.1, 0, 0, 0]

players = ["Sachin", "Sehwag", "Kohli", "Yuvraj"]
runs_5yrs = [1200, 600, 300, 700]

player_colors = ["skyblue", "purple", "yellow", "brown"]

plt.pie(runs_5yrs, labels=players, explode= expl, colors = player_colors, shadow=True)
plt.title("Runs in 5yrs")
plt.show()
```

```python
# startangle : rotating the pie with certain degree

players = ["Sachin", "Sehwag", "Kohli", "Yuvraj"]
runs_5yrs = [1200, 600, 300, 700]

plt.pie(runs_5yrs, labels=players, startangle=-65)
plt.title("Runs in 5yrs")
plt.show()
```

```python
# autopct = '%1.1f%%'

players = ["Sachin", "Sehwag", "Kohli", "Yuvraj"]
runs_5yrs = [1200, 600, 300, 700]

plt.pie(runs_5yrs, labels=players, startangle=-65, autopct = '%1.1f%%')
plt.title("Runs in 5yrs")
plt.show()
```
