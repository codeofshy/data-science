```python
years = [1990, 1992, 1994, 1996, 1998, 2000, 2003, 2005, 2007, 2010]
runs =  [500, 700, 1100, 1500, 1800, 1200, 1700, 1300, 900, 1500]

plt.plot(years, runs)
plt.show()
```

```python
plt.plot(years, runs)
plt.xlabel("Years")
plt.ylabel("Runs")
plt.title("Shachin's Runs Vs Years")
plt.show()
```

```python
years = [1990, 1992, 1994, 1996, 1998, 2000, 2003, 2005, 2007, 2010]
kohli = [0, 0, 500, 800, 1100, 1300, 1500, 1800, 1900, 2100]
sehwag = [0, 300, 800, 1200, 1500, 1700, 1600, 1400, 1000, 0]

plt.plot(years, kohli, label="Kohli's Score")
plt.plot(years, sehwag, label="Sehwag's Score")

plt.xlabel("Years")
plt.ylabel("Runs")
plt.title("Performance card")
plt.legend()
plt.show()
```

```python
# https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.plot.html


# 'ro--', 'g^:' : similarly we can look for other markers in the above URL
plt.plot(years, kohli, 'ro--', label="Kohli")  # red circles with dashed lines
plt.plot(years, sehwag, 'g^:', label="Sehwag")  # green triangles dotted
plt.legend()
plt.show()
```

```python
# formating with better readability

plt.plot(years, kohli, color='orange', linestyle='--', label="Kohli")
plt.plot(years, sehwag, color='green', linestyle='-.', label="Sehwag")
plt.plot(years, runs, color='blue', label="Tendulkar")
plt.legend()
plt.show()
```

```python
plt.plot(years, kohli, color='orange', linestyle='--',linewidth=3, label="Kohli")   # linewidth used to thicken the lines
plt.plot(years, sehwag, color='green', linestyle='-.', label="Sehwag")
plt.plot(years, runs, color='blue', label="Tendulkar")
plt.legend()
plt.show()
```

```python
plt.plot(years, kohli, color='orange', linestyle='--',linewidth=3, label="Kohli")   # linewidth used to thicken the lines
plt.plot(years, sehwag, color='green', linestyle='-.', label="Sehwag")
plt.plot(years, runs, color='blue', label="Tendulkar")
plt.legend()
plt.tight_layout()   # shows the graps in tight packing
plt.show()
```
