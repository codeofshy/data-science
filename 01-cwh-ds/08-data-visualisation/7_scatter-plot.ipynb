```python
import matplotlib.pyplot as plt

# Sample data
study_hours = [1, 2, 3, 4, 5, 6, 7, 8, 9]
exam_scores = [40, 45, 88, 55, 20, 65, 75, 85, 90]

colors = ["red" if score < 50 else "green" for score in exam_scores]
size = [score*2 for score in exam_scores]

plt.scatter(study_hours, exam_scores, c=colors, s=size)
plt.xlabel("Hours")
plt.ylabel("Score")
plt.grid(True)
plt.title("Hours vs Scores")
plt.show()
```

```python
import numpy as np

# Sample data
study_hours = [1, 2, 3, 4, 5, 6, 7, 8, 9]
exam_scores = [40, 45, 88, 55, 20, 65, 75, 85, 90]

# Also works with Numpy Arrays
scores_normalized = np.array(exam_scores)

plt.scatter(study_hours, exam_scores, c=scores_normalized, cmap='viridis')
# the line below is giving us colorbar line
plt.colorbar(label='Score')
plt.title('Scatter Plot with Colormap')
plt.xlabel('Study Hours')
plt.ylabel('Exam Score')
plt.grid(True)
plt.show()
```

```python
# add label
import numpy as np

# Sample data
study_hours = [1, 2, 3, 4, 5, 6, 7, 8, 9]
social_score = [11,23,46,350,200,420,600, 999, 445]

# Also works with Numpy Arrays
# scores_normalized = np.array(exam_scores)

plt.scatter(study_hours, social_score, c=social_score, cmap='viridis')
# the line below is giving us colorbar line
plt.colorbar(label='Score')
for i in range(len(study_hours)):
    plt.annotate(f'S {i}', (study_hours[i], social_score[i]))
plt.title('Scatter Plot with Colormap')
plt.xlabel('Study Hours')
plt.ylabel('Social Score')
plt.grid(True)
plt.show()
```

```python
# Assume two groups: Class A and Class B
class_a_hours = [2, 4, 6, 8]
class_a_scores = [45, 55, 65, 85]

class_b_hours = [1, 3, 5, 7, 9]
class_b_scores = [40, 50, 60, 70, 90]

plt.scatter(class_a_hours, class_a_scores, label='Class A', color='blue')
plt.scatter(class_b_hours, class_b_scores, label='Class B', color='orange')

plt.title('Scatter Plot: Class A vs Class B')
plt.xlabel('Study Hours')
plt.ylabel('Exam Score')
plt.legend()
plt.grid(True)
plt.show()
```
