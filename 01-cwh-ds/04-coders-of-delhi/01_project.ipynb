```python
import json
```

```python
# creating a function to load the data
def load_json(filename):
    with open(filename, 'r') as f:
        data = json.load(f)
    return data
    
```

```python
data = load_json('data.json')
```

```python
type(data)
```

```python
# write a function to display the pages

def display_pages(data):
    for page in data['pages']:
        print(f"{page['id']}. {page['name']}")

display_pages(data)
```

```python
# write a function to display user data and connection
def display_user_data(data):
    print("Users and their connections : ")
    for user in data['users']:
        # print(user)
        print(f"{user['id']}. User name is {user['name']} is friends with {user['friends']} and pages liked are {user['liked_pages']}.")
    print('\n')
    print('Pages')
    # function inside function
    display_pages(data)
        
```

```python
display_user_data(data)
```

```python
content = display_user_data(data)

print(type(content))
```
