```python
import pandas as pd

employees = pd.DataFrame({
    "EmpID": [1, 2, 3],
    "Name": ["Alice", "Bob", "Charlie"],
    "DeptID": [10, 20, 30]
})

departments = pd.DataFrame({
    "DeptID": [10, 20, 40],
    "DeptName": ["HR", "Engineering", "Marketing"]
})
```

```python
employees
```

```python
departments
```

```python
pd.merge(employees, departments, on="DeptID")  # Inner join : just give the common data from both table
```

```python
pd.merge(employees, departments, on="DeptID", how="left") # left join
```

```python
pd.merge(employees, departments, on="DeptID", how="right") # right join
```

```python
pd.merge(employees, departments, on="DeptID", how="outer") # outer join
```

```python
df1 = pd.DataFrame({"Name" : ["Alice", "Bob"], "Score" : [99, 56]})
df2 = pd.DataFrame({"Name" : ["Jack", "Jill"], "Age" : [12, 42]})
```

```python
df1
```

```python
df2
```

```python
pd.concat([df1,df2])
```

```python
pd.concat([df1,df2], axis=1)
```

# when do we need concat
- When we have our data in multiple chunks.
- for example if we have multiple data from multiple csv, url, json, sql.
- we will create a different df and finally we can concat the data.
