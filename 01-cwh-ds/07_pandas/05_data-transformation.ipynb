```python
import pandas as pd

df = pd.read_csv('data.csv')
df
```

```python
# sorting with Year
df.sort_values("Year")
```

```python
# not in descending order
df.sort_values("Year", ascending=False)
```

```python
df2 = df.sort_values(["Year","IMDb"]).copy()
```

```python
#  now as you can see we have mixed index

# this will create a new index and also gives you the old index as a column
df2.reset_index()
```

```python
df2
```

```python
# this will not create a new index column, and inplace will make sure the changes are done to original dataframe as well
df2.reset_index(drop=True, inplace=True)
```

```python
# df2.sort_index()  # this will sort dataframe according to the index
```

```python
df2
```

```python
df2['Rank'] = df2['IMDb'].rank()


df2
```

```python
df2['Rank'] = df2['IMDb'].rank(ascending=False)


df2
```

```python
#  if you want the rank which is TIE..! to have a integer value
df2['Rank'] = df2['IMDb'].rank(ascending=False, method="dense")


df2
```
