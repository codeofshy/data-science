# Python Comprehensions Cheat Sheet

## 1. List Comprehension
### Basic
```python
squares = [num ** 2 for num in range(1, 6)]
print(squares)
```

### Conditional
```python
evens = [num for num in range(1, 11) if num % 2 == 0]
print(evens)
```

### If-Else
```python
labels = ['Even' if num % 2 == 0 else 'Odd' for num in range(1, 6)]
print(labels)
```

## 2. Set Comprehension
### Basic
```python
squares = {num ** 2 for num in range(1, 6)}
```

### Conditional
```python
evens = {num for num in range(1, 11) if num % 2 == 0}
```

### If-Else
```python
labels = {'Even' if num % 2 == 0 else 'Odd' for num in range(1, 6)}
```

## 3. Dictionary Comprehension
### Basic
```python
squares = {num: num ** 2 for num in range(1, 6)}
```

### Conditional
```python
even_squares = {num: num ** 2 for num in range(1, 11) if num % 2 == 0}
```

### If-Else
```python
status = {num: 'Even' if num % 2 == 0 else 'Odd' for num in range(1, 6)}
```

## 4. Generator Comprehension
### Basic
```python
gen = (num ** 2 for num in range(1, 6))
```

### Conditional
```python
gen = (num for num in range(1, 11) if num % 2 == 0)
```

### If-Else
```python
gen = ('Even' if num % 2 == 0 else 'Odd' for num in range(1, 6))
```

## Master Patterns
```python
[expression for item in iterable if condition]
```

```python
[true_value if condition else false_value for item in iterable]
```
