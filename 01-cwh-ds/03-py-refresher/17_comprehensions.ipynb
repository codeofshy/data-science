## comprehensions
There are 4 main comprehension types you should know:
- List Comprehension
- Set Comprehension
- Dictionary Comprehension
- Generator Comprehension

Expression for comprehension : <b>[expression/result</b> for <b>item </b>in<b> iterable]</b>

```python
# normal loop
squares = []

for num in range(1, 6):
    squares.append(num ** 2)

print(squares)
```

```python
list comprehension
```

```python

squares = [num ** 2 for num in range(1, 6)]

print(squares)
```

```python
'''
Memory Trick

Read it left to right:

"Take num for every num if num is even"
'''

evens = [num for num in range(1, 11) if num % 2 == 0]

print(evens)
```

```python
# [value_if_true if condition else value_if_false for item in iterable]
labels = ["Even" if num % 2 == 0 else "Odd"
          for num in range(1, 6)]

print(labels)
```

Set Comprehension

```python
squares = {num ** 2 for num in range(1, 6)}

print(squares)
```

Why Sets?

Sets automatically remove duplicates.

```python
numbers = [1, 2, 2, 3, 3, 4]

unique = {num for num in numbers}

print(unique)
```

```python
labels = {"Even" if num % 2 == 0 else "Odd"
          for num in range(1, 6)}

print(labels)
```

Dictionary comprehension

```python
squares = {num: num ** 2
           for num in range(1, 6)}

print(squares)
```

```python
even_squares = {
    num: num ** 2
    for num in range(1, 11)
    if num % 2 == 0
}

print(even_squares)
```

```python
status = {
    num: "Even" if num % 2 == 0 else "Odd"
    for num in range(1, 6)
}

print(status)
```

```python
gen = (num ** 2 for num in range(1, 6))

print(gen)
```

- [expression for item in iterable if condition]
- [num for num in range(10) if num % 2 == 0]
- [expression_if_true if condition else expression_if_false
 for item in iterable]

- []  -> List Comprehension
- {}  -> Set Comprehension
- {k:v} -> Dictionary Comprehension
- ()  -> Generator Comprehension

- list -> [x for x in items]
- set  -> {x for x in items}
- dicts-> {x: x*x for x in items}
- generator -> (x for x in items)
