## File Handling
- File handling allows Python programs to read, write, and manipulate files stored on disk.
- Python provides built-in functions for working with files.

```python
# to read a file
file = open('file-to-read.txt', 'r')
```

```python
print(file)
```

```python
file.read()
```

```python
file.close
```

```python
file.close()
```

```python
# since the file is closed we cannot read it now
file.read()
```

```python
f = open('file-to-write.txt','w')
```

```python
s = 'This is a nice way to learn python'
```

```python
f.write(s)
```

```python
f.close()
```

```python
with open('file-to-read.txt', 'r') as file:
    content = file.read()
    print(content)
```

```python
with open('text.txt', 'r') as file:
    content = file.read()
    print(content)
```

```python
with open('newText.txt', "w") as f:
    s = 'I am writing this file using with method'
    f.write(s)
```
