## slicing returns a view not a copy. Changes in copy takes affect in original array
- flat_arr is original array.
- b is a variable where i have sliced flat_arr
- now when i modify any element of b
- same element of flat_arr also gets affected

Theory
- This happens because numpy does not give you copy while slicing.
- Instead, it gives you the original view of the memory for the slice.
- As, they share the same memory any modification in sliced item will change in the original array and vice-versa.
- for eficient memory

```python
# Fancy indexing : you cna give multiple index in the form of list

arr = np.array([10, 20, 30, 40, 50])
idx = [0, 2, 4]  # Indices to select
print(arr[idx])  # [10 30 50]
```

```python
# Boolean masking : the condition you give in [], will filter the items of the array

arr = np.array([10, 20, 30, 40, 50])
mask = arr > 25  # Condition: values greater than 25
print(arr[mask])  # [30 40 50]
```
