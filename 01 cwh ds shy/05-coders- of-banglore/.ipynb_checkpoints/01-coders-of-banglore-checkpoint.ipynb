```python
'''
{'user_name': 'intaglobal',
 'no_of_post': '1,946 posts',
 'no_of_followers': '6,851 followers',
 'no_of_following': '262 following',
 'name': 'INTA',
 'type_of_page': 'Nonprofit organization',
 'bio': 'The association of ™ professionals.\ninta.campsite.bio'}
'''


def parse_chunk(chunk):
    chunk = chunk.strip()
    sep_chunk = chunk.split('\n') # ['intaglobal', '1,946 posts', '6,851 followers', '262 following', 'INTA', 'Nonprofit organization', 'The association of ™ professionals.', 'inta.campsite.bio']
    user_name = sep_chunk[0] 
    # no_of_post = sep_chunk[1] 
    # no_of_followers = sep_chunk[2] 
    # no_of_following = sep_chunk[3] 
    no_of_post = int(sep_chunk[1].split(' post')[0].replace(',',''))
    
    no_of_followers = float(sep_chunk[2].split(' followers')[0].replace(',','').replace('K','').replace('M',''))
    if 'K' in (sep_chunk[2]) : 
        no_of_followers *= 1000
    if 'M' in (sep_chunk[2]) : 
        no_of_followers *= 1000_000
        
    no_of_following = float(sep_chunk[3].split(' following')[0].replace(',','').replace('K','').replace('M',''))
    if 'K' in (sep_chunk[3]) : 
        no_of_following *= 1000
    if 'M' in (sep_chunk[3]) : 
        no_of_following *= 1000_000
        
    name = sep_chunk[4]
    if len(sep_chunk) > 5:
        type_of_page = sep_chunk[5]
        bio = '\n'.join(sep_chunk[6:])
    else : 
        type_of_page = 'Unknown'
        bio = 'Unknown'
        
    # print(user_name, no_of_post, no_of_followers, no_of_following, name, type_of_page, bio, sep='\n')
    return {'user_name':user_name, 'no_of_post':no_of_post, 'no_of_followers':int(no_of_followers), 'no_of_following':int(no_of_following), 'name':name, 'type_of_page':type_of_page, 'bio':bio}



parse_chunk(chunks[1])
```

```python
all_chunks = []

for chunk in chunks:
    parsed_chunk = parse_chunk(chunk)
    all_chunks.append(parsed_chunk)

# print(all_chunks)
```

```python
import json
s = json.dumps(all_chunks, indent=4)

print(s)
```

```python
len(all_chunks)
```

```python
# who has maximum post
max_post = 0
chunk_max_post = ''
for user in all_chunks:
    if user['no_of_post'] > max_post:
        max_post = user['no_of_post']
        chunk_max_post = user

print(max_post, chunk_max_post, sep='\n')
```

```python
# who has maximum followers
max_followers = 0
chunk_max_followers = ''
for user in all_chunks:
    if user['no_of_followers'] > max_followers:
        max_followers = user['no_of_followers']
        chunk_max_followers = user

print(max_followers, chunk_max_followers, sep='\n')
```

```python
# who has maximum followers
max_following = 0
chunk_max_following = ''
for user in all_chunks:
    if user['no_of_following'] > max_following:
        max_following = user['no_of_following']
        chunk_max_following = user

print(max_following, chunk_max_following, sep='\n')
```

```python
# how many categories
categories = []
for user in all_chunks:
    category = user['type_of_page']
    categories.append(category)

print(len(categories))
print(categories)
```
