```python
df.iloc[0] # from iloc we get the data by normal indexing like 0, 1, 2
```

```python
df.loc[0, 'IMDb'] # by labels
```

```python
df.iloc[0, 5] # by indexing
```

```python
# loc is inclusive of slicing | 0:3 gives 4 rows from 0 to 3
df.loc[0:3, ['Actor', 'IMDb', 'Film']]
```

```python
# iloc is exclusive of slicing | 0:3 gives 3 rows from 0 to 2
df.iloc[0:3, [0, 1, 5]]
```
