```python
# we can also change the data type of a column
# here the age is in float, converting it with int
df_clean = df.dropna().copy()
```

```python
df_clean
```

```python
df_clean['Age'] = df_clean['Age'].astype(int)
```

```python
df_clean
```

## apply()

```python
# using apply to create a column name Age Group
df_clean['Age Group'] = df_clean['Age'].apply(lambda x : 'Young' if x < 25 else 'Old')
```

```python
df_clean
```

```python
#  we can to the same using function

def is_minor(age):
    return 'Adult' if age > 25 else "Minor"

df_clean['Age group def func'] = df_clean['Age'].apply(is_minor)
```

```python
df_clean
```

## map()

```python
gender_map = {"M" : "Male", "F" : "Female", "O" : "Other"}

df_clean["Gender"] = df_clean["Gender"].map(gender_map)

df_clean
```

```python
df_clean["City"] = df_clean["City"].replace({"New York": "NY-city", "Mumbai": "Bombay"})
```
