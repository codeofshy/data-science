```python
import time
l1 = list(range(5000))

# python list - loops
start_list = time.time()
l2 = [x**2 for x in l1]
end_list = time.time()
print('Time taken by list ', (end_list-start_list))
# print(l2)

# numpy array - vectorization
l3 = np.array(l1)
start_np = time.time()
l3 = l3 ** 2
end_np = time.time()
print('Time taken by array : ',  end_np - start_np)
# print(l3)
```
