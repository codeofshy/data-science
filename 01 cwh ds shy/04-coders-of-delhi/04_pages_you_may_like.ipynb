```python
import json

# func to load the data
def display_data(filename):
    with open(filename, 'r') as f:
        return json.load(f)

# func for pages-you-may-like
def pages_you_may_like(user_id, data):
    user_pages = {}   # {1: {101}, 2: {102}, 3: {101, 103}, 4: {104}}

    for user in data['users']:
        user_pages[user['id']] = set(user['liked_pages'])

    if user_id not in user_pages:
        return 'Unknown user-id...!'

    # user_iked_pages = {}
    # for user in data['users']:
    #     if user['id'] == user_id:
    #         liked_pages[user_id] = user['liked_pages']

    user_liked_pages = user_pages[user_id]

    page_suggestion = {}

    for other_user, pages in user_pages.items():
        if other_user != user_id:
            shared_pages = user_liked_pages.intersection(pages)
        for page in pages:
            if page not in user_liked_pages:
                page_suggestion[page] = page_suggestion.get(page, 0) + len(shared_pages)

    sorted_pages = sorted(page_suggestion.items(), key = lambda x : x[1], reverse=True)
    
    # return [page_id for page_id, score in sorted_pages]   # 
    # using _ as we don't have to use variable so just giving a simple var name "_"
    return [page_id for page_id, _ in sorted_pages]     


# call to display the data, filename as argument.
data = display_data('massive_data.json')

pages_you_may_like(1, data)
```
