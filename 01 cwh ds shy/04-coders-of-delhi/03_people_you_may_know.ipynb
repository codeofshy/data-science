```python
import json


# function to load the data
def load_data(filename):
    with open(filename, 'r') as f:
        data = json.load(f)
    return data


def people_you_may_know(user_id, data):
    # dict for id and friends
    # this is to get all user id as keys and there direct friends as values
    # {1: [2, 3], 2: [1, 4], 3: [1], 4: [2]}
    user_friends = {} 

    for user in data['users']:
        user_friends[user['id']] = list(set(user['friends']))
        
    # for data sanitization, if the user_id is not present in the data return none.
    if user_id not in user_friends:
        return 'Invalid user id...!'

    # direct friends of the give user_id
    direct_friends = user_friends[user_id]

    suggestion = {}
    # for user id 1 : [2, 3] are friends
    for friend in direct_friends:
        # get the friends for user_id 2 and then for 3
        for mutual in user_friends[friend]:
            # make sure the new user_id which is the friend of 2 is not the user_id we passed, (new_user_id shoul != user_id) 
            if mutual!=user_id and mutual not in direct_friends:
                # sugesstion[5] = suggestion.get(5, 0) + 1
                # if you get the value of 5 and it is 1 > then, 1 + 1 {5 : 2}
                # if you did not get the value of 5 then, initiate 5 as key and 1 as value {5 : 1}
                suggestion[mutual] = suggestion.get(mutual, 0) + 1
    sorted_suggestion = sorted(suggestion.items() ,key= lambda x:x[1], reverse=True)
    # sorted_suggestion this will give me result as tupple in list [(user_id, mutual_score_count)] here for 1, ([4, 1])
    
    # return the list comprehension
    return [user_id for user_id,mutual_score_count in sorted_suggestion]

# load the data
data = load_data('massive_data.json')
user_id = 1
recc = people_you_may_know(user_id, data)
print(recc)
    
```
