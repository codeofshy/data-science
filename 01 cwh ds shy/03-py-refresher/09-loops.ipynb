## loops
1. for loop : used to iterate.
2. while loop : runs untill the condition is while becomes false


```python
for char in 'shayan':
    print(char)
```

```python
# range(0, n-1, step)
for n in range(10):
    print(n)
```

```python
# while loop
count = 0
while count < 3:
    print(count)
    count += 1
```

```python
n = 1
while True:
    print(n)
    if n == 5:
        break
    n += 1
```

- break → Exits the loop.
- continue → Skips to the next iteration.
- pass → Does nothing (used as a placeholder).

```python
for i in range(5):
    if (i==3):
        break
    print(i)
    
```

```python
for i in range(5):
    if (i==3):
        continue
    print(i)
```

```python
for i in range(6):
    pass

print('for loop is not causing error') 
```
