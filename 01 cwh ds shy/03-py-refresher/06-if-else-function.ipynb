## Conditional formatting
- if
- if-else
- if-elif-else

```python
age = 5
```

```python
if age <= 18:
    print('Major')
else :
    print('Minor')
```

```python
if True :
    print('condition is true')
```

```python
if False :
    print('condition is false, Therefor it will not run.')
```

```python
percentage = int(input('Enter your percentage : '))

if percentage >= 80:
    print('First class')
elif percentage >= 60 :
    print('Good, can do better')
else:
    print('Need to work on discipline and focus.')
```

## functions in python


```python
# using def create a function
def greet():
    print('Hello, good morning!')
```

```python
# call the function to execute
greet()
```

```python
def add(a,b):
    return a + b
```

```python
add(3, 3)
```

```python
# default parameter
def say_hello(name = 'User'):
    print(f'Hello {name}, have a great day ahead.')
```

```python
say_hello()
```

```python
say_hello('Shy')
```

## lambda function
- single line function

```python
sqaure = lambda num : num**2
```

```python
sqaure(4)
```

```python
import pandas as pd

data = {'Name': ['Alice', 'Bob', 'Charlie'], 'Score': [85, 40, 75]}

df = pd.DataFrame(data)

df['Result'] = df['Score'].apply(lambda x: 'Pass' if x >= 50 else 'Fail')

print(df)
```
