## Json module
- JSON (JavaScript Object Notation) is a lightweight data format used for data exchange between servers and applications.
- It is widely used in APIs, web applications, and configurations.

```python
import json
```

```python
data = {"name": "Alice", "age": 25, "city": "New York"}
```

- json.dumps : convert python object(here dictionary object) in to a string
- json.loads : convert string into the python object

```python
# dumps : dump as string
json_string = json.dumps(data)
```

```python
json_string
```

```python
type(json_string)
```

```python
json_data = json.loads(json_string)
```

```python
json_data
```

```python
type(json_data)
```

```python
with open('json_data.txt', 'w') as file:
    file.write(str(json_data))
```

```python
# this is used to dump the file into json format
with open('data.json', 'w') as file:
    json.dump(json_data, file)
```

```python
# this will simply write the file in str format and not a json
with open('json_data.json', 'w') as file:
    file.write(str(json_data))
```

```python
with open('data1.json', 'w') as file:
    json.dump(json_data, file, indent=4)
```
