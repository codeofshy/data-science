## Operators precedence
- In which order the operators will be executed.
- PEMDAS - Paranthesis Exponential Mumltiplication Division Addidtion Subtraction
- Exponentials are preferenced from left to right : 3 ** 2 ** 2
- Multiplication and division are preferenced from right to left : 2*3/2 + 1

```python
result = 10 + 2 * 3  # Multiplication happens first: 10 + (2 * 3) = 16
print(result)

result = (10 + 2) * 3  # Parentheses first: (10 + 2) * 3 = 36
print(result)

result = 2 ** 3 ** 2  # Right-to-left exponentiation: 2 ** (3 ** 2) = 2 ** 9 = 512
print(result)
```
