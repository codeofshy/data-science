```python
def http_status(code):
    match code:
        case 200:
            return "OK"
        case 400:
            return "Bad Request"
        case 404:
            return "Not Found"
        case 500:
            return "Internal Server Error"
        case _:
            return "Unknown Status"

print(http_status(200))  # Output: OK
print(http_status(404))  # Output: Not Found
```

```python
def fan(switch = 'off'):
    match switch:
        case 'on':
            print('Fan is ON, grr. grr.. grr...')
        case 'off':
            print('Fan is off, psst...')

fan()
```

```python
fan('on')
```

```python
age = 1

match age:
    case 18:
        print('person is major.')
    case 60:
        print('senior citizen')
    case _: # _ underscore works as else condition
        print('Nothing matched, else condition executed')
```

```python
point = (3, 4)

match point:
    case (0, 0):
        print("Origin")
    case (x, 0):
        print(f"X-Axis at {x}")
    case (0, y):
        print(f"Y-Axis at {y}")
    case (x, y):
        print(f"Point at ({x}, {y})")
```
