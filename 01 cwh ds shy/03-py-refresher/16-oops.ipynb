## class
- Object-Oriented Programming (OOP) is a programming paradigm that organizes code into objects that contain both data (attributes) and behavior (methods).

```python
class Employee:
    company = 'Apple' # class attributes

    def __init__(self, name, salary) : # __init__(self) is a constructor which can take the user input
        self.name = name # instance attribute
        self.salary = salary

    def company_details(self): # class method
        # we used the self.company but company is class attribute and not instance attribute.
        # 1st code searches in instance if not found goes to class and gives us the variable
        print(f'Hello, the company name is {self.company}.')

    def employee_details(self):
        print(f'Name : {self.name}\nSalary : ${self.salary}')

    # this indicates that the method can run independently and there is no need of passing self parameter.
    # as static method is a standalone method
    @staticmethod 
    def print_time():
        print('The time is now!')

    @classmethod # this will only print the class details
    def print_class_details(cls):
        print(f"The company is {cls.company}")
        # print(f"{self.name}") this gave us the error
```

```python
e1 = Employee('Shy', '980_000')
```

```python
e1.company_details()
```

```python
e1.employee_details()
```

```python
e1.print_time()
```

```python
e1.print_class_details()
```
